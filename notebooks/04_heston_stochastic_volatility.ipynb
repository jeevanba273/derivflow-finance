# Heston Stochastic Volatility Model

The Heston model adds a mean-reverting stochastic variance. `v0` and `theta` are variances: 0.04 is a 20% volatility. The Feller condition 2 kappa theta > sigma^2 holds for these parameters.

```python
from derivflow import HestonModel, black_scholes_price, implied_volatility

model = HestonModel()
model.set_parameters(v0=0.04, kappa=2.0, theta=0.04, sigma=0.35, rho=-0.7)
```

## Analytic price and the implied volatility smile

```python
for K in (80, 90, 100, 110, 120):
    price = model.analytic_price(100, K, 1.0, 0.03, "call")
    print(
        f"K={K}: price {price:.4f}  implied vol {implied_volatility(price, 100, K, 1.0, 0.03):.4f}"
    )
```

## Monte Carlo check (QE scheme) and Black-Scholes for comparison

```python
mc = model.monte_carlo_price(
    100, 100, 1.0, 0.03, "call", num_sims=100_000, num_steps=100, random_seed=1
)
print("analytic     :", round(model.analytic_price(100, 100, 1.0, 0.03, "call"), 4))
print("Monte Carlo  :", round(mc.price, 4), "+-", round(mc.std_error, 4))
print("Black-Scholes:", round(black_scholes_price(100, 100, 1.0, 0.03, 0.2), 4))
```

With rho < 0 the model produces a downward-sloping smile, so the at-the-money price is below the flat 20% Black-Scholes value.
